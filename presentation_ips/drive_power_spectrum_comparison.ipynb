# Low-Drive vs High-Drive Spectroscopy

This notebook creates a presentation-ready figure with two stacked spectroscopy plots:

- **Low drive:** narrow T2-limited resonance, but low signal-to-noise ratio.
- **High drive:** higher signal-to-noise ratio, but broader power-broadened resonance.

The output is saved to `../figs/drive_power_spectrum_comparison.png`.

```python
import os
from pathlib import Path
import numpy as np

# Keep Matplotlib cache files in a writable temporary directory.
os.environ.setdefault("MPLCONFIGDIR", "/private/tmp/ips-matplotlib-cache")

import matplotlib
matplotlib.use("Agg")
import matplotlib.pyplot as plt

# Reproducible noise for a stable presentation figure.
rng = np.random.default_rng(7)

# Detuning axis in kHz around resonance.
detuning_khz = np.linspace(-180, 180, 1200)

def lorentzian(x, center, fwhm, amplitude, offset=0.0):
    gamma = fwhm / 2.0
    return offset + amplitude * gamma**2 / ((x - center)**2 + gamma**2)

# Low drive: narrow, T2-limited linewidth but low SNR.
t2_limited_fwhm_khz = 20.0
low_drive_clean = lorentzian(
    detuning_khz,
    center=0.0,
    fwhm=t2_limited_fwhm_khz,
    amplitude=0.23,
    offset=0.04,
)
low_drive_signal = low_drive_clean + rng.normal(0.0, 0.035, size=detuning_khz.size)

# High drive: strong signal, but power-broadened linewidth.
power_broadened_fwhm_khz = 105.0
high_drive_clean = lorentzian(
    detuning_khz,
    center=0.0,
    fwhm=power_broadened_fwhm_khz,
    amplitude=0.86,
    offset=0.04,
)
high_drive_signal = high_drive_clean + rng.normal(0.0, 0.018, size=detuning_khz.size)

# Presentation styling.
plt.rcParams.update({
    "font.size": 22,
    "axes.labelsize": 26,
    "axes.titlesize": 28,
    "xtick.labelsize": 20,
    "ytick.labelsize": 20,
    "legend.fontsize": 19,
    "figure.titlesize": 30,
    "axes.linewidth": 2.0,
})

fig, axes = plt.subplots(1, 2, figsize=(15.5, 6.6), dpi=180, sharey=True)
fig.subplots_adjust(top=0.80, bottom=0.18, left=0.10, right=0.98, wspace=0.16)

# Low-drive plot.
axes[0].plot(detuning_khz, low_drive_signal, color="#315f9f", linewidth=2.2, alpha=0.95)
axes[0].plot(detuning_khz, low_drive_clean, color="#0d2f66", linewidth=4.0)
axes[0].axvline(0, color="black", linestyle="--", linewidth=2.0, alpha=0.65)
axes[0].annotate(
    "T2-limited\nnarrow line",
    xy=(0, low_drive_clean.max()),
    xytext=(52, 0.16),
    arrowprops={"arrowstyle": "->", "linewidth": 2.5, "color": "#0d2f66"},
    fontsize=23,
    color="#0d2f66",
    weight="bold",
)
axes[0].text(-170, 0.86, "Low drive", fontsize=28, weight="bold", color="#0d2f66")
axes[0].text(-170, 0.76, "low SNR", fontsize=24, color="#0d2f66")
axes[0].set_title("Weak spectroscopy drive", pad=14)
axes[0].set_xlabel("Detuning from resonance (kHz)")
axes[0].set_ylabel("Excited-state\npopulation")
axes[0].set_ylim(0.0, 1.0)
axes[0].grid(True, alpha=0.25, linewidth=1.2)

# High-drive plot.
axes[1].plot(detuning_khz, high_drive_signal, color="#b24a16", linewidth=2.2, alpha=0.95)
axes[1].plot(detuning_khz, high_drive_clean, color="#7a2508", linewidth=4.0)
axes[1].axvline(0, color="black", linestyle="--", linewidth=2.0, alpha=0.65)
axes[1].annotate(
    "Power-broadened\nwide line",
    xy=(55, lorentzian(55, 0, power_broadened_fwhm_khz, 0.86, 0.04)),
    xytext=(82, 0.56),
    arrowprops={"arrowstyle": "->", "linewidth": 2.5, "color": "#7a2508"},
    fontsize=23,
    color="#7a2508",
    weight="bold",
)
axes[1].text(-170, 0.86, "High drive", fontsize=28, weight="bold", color="#7a2508")
axes[1].text(-170, 0.76, "high SNR", fontsize=24, color="#7a2508")
axes[1].set_title("Strong spectroscopy drive", pad=14)
axes[1].set_xlabel("Detuning from resonance (kHz)")
axes[1].set_ylim(0.0, 1.0)
axes[1].grid(True, alpha=0.25, linewidth=1.2)

for ax in axes:
    ax.tick_params(width=2.0, length=7)
    ax.spines["top"].set_visible(False)
    ax.spines["right"].set_visible(False)

fig.suptitle("Conventional Spectroscopy Trade-off", y=0.97, weight="bold")

cwd = Path.cwd()

project_root = cwd.parent if cwd.name == "codes" else cwd
output_dir = project_root / "figs"
output_dir.mkdir(parents=True, exist_ok=True)
output_path = output_dir / "drive_power_spectrum_comparison.png"
fig.savefig(output_path, bbox_inches="tight", facecolor="white")
plt.show()

print(f"Saved figure to {output_path}")
```
