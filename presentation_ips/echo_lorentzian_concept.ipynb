# Echo-Root-Lorentzian Pulse Concept

This notebook creates a presentation-ready conceptual figure for the opening slide.

It shows the root-Lorentzian pulse envelope, the pi phase inversion at the pulse center, and the resonance-cancellation intuition.

The output is saved to `figs/echo_lorentzian_concept.png`.

```python
import os
from pathlib import Path

import numpy as np

# Keep Matplotlib cache files in a writable temporary directory.
os.environ.setdefault("MPLCONFIGDIR", "/private/tmp/ips-matplotlib-cache")

import matplotlib
matplotlib.use("Agg")
import matplotlib.pyplot as plt

# Time axis in units of the Lorentzian timescale tau.
t = np.linspace(-5.0, 5.0, 2000)
tau = 1.0

# Root-Lorentzian envelope and echo version with a pi phase flip at t = 0.
root_lorentzian = 1.0 / np.sqrt(1.0 + (t / tau) ** 2)
echo_root_lorentzian = root_lorentzian * np.sign(t)
echo_root_lorentzian[t == 0] = 0.0

# Smooth color regions for the two pulse halves.
left = t < 0
right = t > 0

plt.rcParams.update({
    "font.size": 22,
    "axes.labelsize": 25,
    "axes.titlesize": 28,
    "xtick.labelsize": 18,
    "ytick.labelsize": 18,
    "figure.titlesize": 34,
    "axes.linewidth": 2.0,
})

fig, ax = plt.subplots(figsize=(15.5, 7.2), dpi=180)
fig.subplots_adjust(top=0.82, bottom=0.18, left=0.09, right=0.98)

# Envelope and signed echo pulse.
ax.plot(t, root_lorentzian, color="#6f6f6f", linewidth=3.2, linestyle="--", alpha=0.85)
ax.plot(t[left], echo_root_lorentzian[left], color="#1f6f8b", linewidth=5.0)
ax.plot(t[right], echo_root_lorentzian[right], color="#c4472d", linewidth=5.0)
ax.fill_between(t[left], 0, echo_root_lorentzian[left], color="#1f6f8b", alpha=0.16)
ax.fill_between(t[right], 0, echo_root_lorentzian[right], color="#c4472d", alpha=0.16)

# Central phase jump.
ax.axvline(0, color="black", linestyle="--", linewidth=2.5, alpha=0.75)
ax.scatter([0], [0], s=170, color="black", zorder=5)
ax.text(0.18, 1.02, r"$\pi$ phase flip", fontsize=28, weight="bold", color="black")

# Annotate the two halves.
ax.text(-4.65, -0.82, "first half", fontsize=28, weight="bold", color="#1f6f8b")
ax.text(-4.65, -1.02, r"phase $0$", fontsize=23, color="#1f6f8b")
ax.text(2.65, 0.82, "second half", fontsize=28, weight="bold", color="#c4472d")
ax.text(2.65, 0.62, r"phase $\pi$", fontsize=23, color="#c4472d")

# Cancellation arrows around the center.
ax.annotate(
    "resonant evolution",
    xy=(-1.0, -0.72),
    xytext=(-3.25, -0.38),
    arrowprops={"arrowstyle": "->", "linewidth": 3.0, "color": "#1f6f8b"},
    fontsize=24,
    color="#1f6f8b",
)
ax.annotate(
    "reversed evolution",
    xy=(1.0, 0.72),
    xytext=(1.95, 0.36),
    arrowprops={"arrowstyle": "->", "linewidth": 3.0, "color": "#c4472d"},
    fontsize=24,
    color="#c4472d",
)
ax.annotate(
    "on resonance:\nnet rotation cancels",
    xy=(0.0, 0.0),
    xytext=(-1.3, 1.22),
    arrowprops={"arrowstyle": "->", "linewidth": 3.0, "color": "black"},
    fontsize=27,
    color="black",
    weight="bold",
    ha="center",
)

# Add formula directly on the figure.
ax.text(
    -4.75,
    1.22,
    r"$L_{echo}^{1/2}(t)=\frac{\mathrm{sgn}(t)}{\sqrt{1+(t/\tau)^2}}$",
    fontsize=30,
    color="#222222",
)

ax.set_title("Echo-Root-Lorentzian Pulse: Strong Drive Without Resonant Rabi Build-up", pad=18)
ax.set_xlabel(r"Time relative to pulse center ($t/\tau$)")
ax.set_ylabel("Signed drive\namplitude")
ax.set_xlim(-5, 5)
ax.set_ylim(-1.18, 1.45)
ax.set_yticks([-1, 0, 1])
ax.grid(True, alpha=0.22, linewidth=1.2)
ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)
ax.tick_params(width=2.0, length=7)

cwd = Path.cwd()
project_root = cwd.parent if cwd.name == "codes" else cwd
output_dir = project_root / "figs"
output_dir.mkdir(parents=True, exist_ok=True)
output_path = output_dir / "echo_lorentzian_concept.png"
fig.savefig(output_path, bbox_inches="tight", facecolor="white")
plt.show()

print(f"Saved figure to {output_path}")
```
