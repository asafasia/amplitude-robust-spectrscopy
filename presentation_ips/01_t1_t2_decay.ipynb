# IPS T1/T2 Decay Figure

Presentation-ready T1/T2 decay figure using the project figure workflow.

```python
from echospec.figures import FigureVariant, apply_figure_style, save_figure
from echospec.figures.decay import make_t1_t2_decay_figure

VARIANT = FigureVariant.PRESENTATION_IPS
apply_figure_style(VARIANT)
```

```python
fig = make_t1_t2_decay_figure(
    t1_us=30.0,
    t2_us=14.0,
    variant=VARIANT,
)

saved_paths = save_figure(
    fig,
    "01_t1_t2_decay_first_slide",
    variant=VARIANT,
    formats=("png", "svg"),
)
saved_paths
```
